### Задание 6.1.3. Функция flatten_list: создание, документирование, тестирование

```python
def flatten_list(nested_obj, gen: bool=False):
    '''
    Рекурсивная функция, выравнивающая структуру списка nested_obj.

    Аргументы:

    nested_obj: list/non-interable - итеративный либо неитеративный объект,
    структуру которого нужно выровнять.

    gen: bool = False - вернёт ли функция генеративный объект; по умолчанию False -
    функция возвращает сам список
    '''
    if gen:
        if isinstance(nested_obj,list):
            for item in nested_obj:
                for i in flatten_list(item, gen):
                    yield i
        else:
            yield nested_obj
    else:
        result = []
        if isinstance(nested_obj,list):
            for item in nested_obj:
                result += flatten_list_v1(item) # рекурсия
        else:
            result += [nested_obj]
        return result
```

### Задание 6.2.1. Рекурсивная функция flatten_it_v1

```python
def flatten_it_v1(nested_obj):
    result = []
    if hasattr(nested_obj, "__iter__"):
        if isinstance(nested_obj, str):  # Если строка - не итерируем
            result += [nested_obj]
            return result                # Сразу возвращаем результат
        if isinstance(nested_obj,dict):  # Если словарь
            nested_obj = nested_obj.items() # то берём вместо ключей
                                            # пары (ключ-значение)
        for i in nested_obj:
            result += flatten_it_v1(i)   # Рекурсивный вызов
    else:
        result += [nested_obj]
    return result
```

```python
[flatten_it_v1(e) for e in it]
```

### Задание 6.2.2. Генераторная функция на основе рекурсии flatten_it_v2

```python
def flatten_it_v2(nested_obj):
    if hasattr(nested_obj, "__iter__"):
        if isinstance(nested_obj, str):  # Если строка - не итерируем
            yield nested_obj
            return                      # Сразу возвращаем результат
        if isinstance(nested_obj,dict):  # Если словарь
            nested_obj = nested_obj.items() # то берём вместо ключей
                                            # пары (ключ-значение)
        for i in nested_obj:
            for e in flatten_it_v2(i):
                yield e                  # Рекурсивный вызов
    else:
        yield nested_obj
```

### Задание 6.2.3. Функция flatten_it: создание, документирование, тестирование

```python
def flatten_it(nested_obj, gen = False):
    '''
    Функция flatten_it(nested_obj, gen) выравнивает структуру
    итерируемого объекта, возвращая либо список элементов,
    либо генераторный объект элементов

    Аргументы:

    nested_obj - итерируемый объект. Может быть любым встроенным
    типом данных.

    gen: bool = False - будет ли возвращаемый объект генератором
    '''
    if gen:
        def generator(nested_obj):
            if hasattr(nested_obj, "__iter__"):
                if isinstance(nested_obj, str):
                    yield nested_obj
                    return
                if isinstance(nested_obj,dict):
                    nested_obj = nested_obj.items()
                for i in nested_obj:
                    for e in flatten_it(i, True):
                        yield e
            else:
                yield nested_obj
        return generator(nested_obj)
    else:
        result = []
        if hasattr(nested_obj, "__iter__"):
            if isinstance(nested_obj, str):
                result += [nested_obj]
                return result
            if isinstance(nested_obj,dict):
                nested_obj = nested_obj.items()
            for i in nested_obj:
                result += flatten_it(i)
        else:
            result += [nested_obj]
        return result
```

### Задание 6.3.1. Доопределение функции flatten_it

```python
def flatten_it(nested_obj, gen = False):
    '''
    Функция flatten_it(nested_obj, gen) выравнивает структуру
    итерируемого объекта, возвращая либо список элементов,
    либо генераторный объект элементов

    Аргументы:

    nested_obj - итерируемый объект. Может быть любым встроенным
    типом данных.

    gen: bool = False - будет ли возвращаемый объект генератором

    Если в функции в качестве аргумента nested_obj указать циклический
    объект, то вызовется ошибка ValueError
    '''
    if isinstance(nested_obj, dict):
        if nested_obj in nested_obj.values():
            raise ValueError("Cyclic dict is found")
    if gen:
        def generator(nested_obj):
            if hasattr(nested_obj, "__iter__"):
                if isinstance(nested_obj, str):
                    yield nested_obj
                    return
                if isinstance(nested_obj,dict):
                    nested_obj = nested_obj.items()
                for i in nested_obj:
                    if i is nested_obj:
                        raise ValueError("Cyclic object is found")
                    for e in flatten_it(i, True):
                        yield e
            else:
                yield nested_obj
        return generator(nested_obj)
    else:
        result = []
        if hasattr(nested_obj, "__iter__"):
            if isinstance(nested_obj, str):
                result += [nested_obj]
                return result
            if isinstance(nested_obj,dict):
                nested_obj = nested_obj.items()
            for i in nested_obj:
                if i is nested_obj:
                    raise ValueError("Cyclic object is found")
                result += flatten_it(i)
        else:
            result += [nested_obj]
        return result
```
