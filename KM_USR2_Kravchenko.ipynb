### Краткая теория

`__init__` — специальный метод инициализации объекта. Он автоматически вызывается после создания нового экземпляра класса и обычно задаёт начальные значения атрибутов объекта.

`__str__` — специальный метод строкового представления объекта. Он вызывается, когда объект передают в `print()` или преобразуют через `str()`.

`__del__` — специальный метод, который вызывается перед уничтожением объекта сборщиком мусора. В учебных примерах его можно использовать для демонстрации удаления объекта, но в реальных программах на него обычно не полагаются, потому что момент вызова может зависеть от среды выполнения.

**Атрибут класса** принадлежит самому классу и общий для всех объектов, если его не переопределить у конкретного экземпляра.

**Атрибут экземпляра** принадлежит конкретному объекту и хранит индивидуальные данные именно этого объекта.

```python
class StudentDemo:
    university = "БГУ"          # атрибут класса: общий для всех студентов
    students_count = 0           # атрибут класса: общий счётчик созданных объектов

    def __init__(self, name, course):
        self.name = name         # атрибут экземпляра: имя конкретного студента
        self.course = course     # атрибут экземпляра: курс конкретного студента
        StudentDemo.students_count += 1

    def __str__(self):
        return f"{self.name}, {self.course} курс, университет: {StudentDemo.university}"

    def __del__(self):
        print(f"Объект студента {self.name} удаляется")


student1 = StudentDemo("Сергей", 1)
student2 = StudentDemo("Анна", 2)

print(student1)
print(student2)
print("Количество созданных студентов:", StudentDemo.students_count)
print("Университет student1:", student1.university)
print("Университет student2:", student2.university)

del student2
```

## Задание 2


- создать экземпляры каждого класса;
- получить список суперклассов каждого класса;
- определить, к какому классу относится каждый объект;
- изобразить дерево наследования.

```python
class Entity:
    def __init__(self, name):
        self.name = name

    def describe(self):
        return f"Объект: {self.name}"


class Person(Entity):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age

    def describe(self):
        return f"Человек: {self.name}, возраст: {self.age}"


class Robot:
    def __init__(self, model):
        self.model = model

    def system_info(self):
        return f"Модель робота: {self.model}"


class Cyborg(Person, Robot):
    def __init__(self, name, age, model):
        Person.__init__(self, name, age)
        Robot.__init__(self, model)

    def describe(self):
        return f"Киборг: {self.name}, возраст: {self.age}, модель: {self.model}"


entity = Entity("Базовая сущность")
person = Person("Сергей", 18)
robot = Robot("R2")
cyborg = Cyborg("Алекс", 25, "Cyber-X")

objects = [entity, person, robot, cyborg]
classes = [Entity, Person, Robot, Cyborg]

print("Объекты и их классы:")
for obj in objects:
    print(obj, "->", type(obj).__name__)

print("\nСуперклассы:")
for cls in classes:
    print(cls.__name__, "->", [base.__name__ for base in cls.__bases__])

print("\nДерево наследования:")
print("Entity")
print("└── Person")
print("    └── Cyborg")
print("Robot")
print("└── Cyborg")
```

## Задание 3



Класс должен содержать:
- метод инициализации координат `x` и `y`;
- метод вычисления длины вектора;
- перегрузку операции сложения `+`;
- перегрузку операции сравнения `==`;
- демонстрацию работы всех методов.

```python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def length(self):
        return (self.x ** 2 + self.y ** 2) ** 0.5

    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __str__(self):
        return f"Vector2D({self.x}, {self.y})"


v1 = Vector2D(3, 4)
v2 = Vector2D(1, 2)
v3 = Vector2D(3, 4)

print("v1 =", v1)
print("v2 =", v2)
print("Длина v1:", v1.length())
print("v1 + v2 =", v1 + v2)
print("v1 == v2:", v1 == v2)
print("v1 == v3:", v1 == v3)
```

## Задание 4
 Для работы со структурой данных очередь определить класс `Queue` на основе встроенного класса `list`.

Класс должен содержать:
- метод `enqueue()` для добавления элемента;
- метод `dequeue()` для удаления элемента из начала очереди.

Также нужно создать экземпляр класса и добавить в очередь элементы, полученные функцией `range(50)`.

```python
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if len(self.items) == 0:
            return None
        return self.items.pop(0)

    def __str__(self):
        return str(self.items)


q = Queue()

for number in range(50):
    q.enqueue(number)

print("Очередь после добавления range(50):")
print(q)

print("Удаляем первые 5 элементов:")
for i in range(5):
    print(q.dequeue())

print("Очередь после удаления 5 элементов:")
print(q)
```

### Пояснение терминов к заданию 5

**Бинарное дерево** — структура данных, где каждый узел может иметь не больше двух потомков: левый и правый.

**Узел дерева** — отдельный элемент дерева, который хранит значение и ссылки на потомков.

**Корень дерева** — самый верхний узел дерева.

**Композиция** — отношение «часть–целое». В этой программе дерево состоит из узлов, а узлы содержат другие узлы или пустые узлы. Поэтому `Tree`, `TreeNode` и `NullNode` связаны композицией.

**Почему нельзя добавить `True` и `None` в данной реализации:** дерево строится для списка целых чисел. В Python `True` технически связан с типом `bool`, а `bool` является наследником `int`, но для строгого учебного дерева целых чисел мы проверяем именно `type(value) is int`, поэтому `True` не добавляется. `None` также не добавляется, потому что это не число и его нельзя корректно сравнивать с целыми числами.

## Итоговая терминология

**Класс** — шаблон для создания объектов.

**Объект** — конкретный экземпляр класса.

**Атрибут** — переменная внутри класса или объекта.

**Метод** — функция, определённая внутри класса.

**`self`** — ссылка на текущий объект. Через `self` объект обращается к своим атрибутам и методам.

**Атрибут класса** — общий атрибут, принадлежащий классу.

**Атрибут экземпляра** — атрибут, принадлежащий конкретному объекту.

**Инкапсуляция** — принцип ООП, при котором данные и методы объединяются внутри класса, а доступ к внутренним данным может быть ограничен.

**Наследование** — возможность создавать новый класс на основе существующего.

**Множественное наследование** — наследование сразу от нескольких классов.

**Перегрузка операторов** — определение поведения стандартных операторов, например `+` или `==`, для объектов пользовательских классов.

**Композиция** — отношение, при котором один объект состоит из других объектов.

**Свойство `property`** — способ работать с методом как с обычным атрибутом, сохраняя контроль над чтением и изменением значения.

**Статический метод** — метод, который логически связан с классом, но не использует ни объект, ни сам класс.

**Классовый метод** — метод, который получает класс через параметр `cls` и может изменять состояние класса.
