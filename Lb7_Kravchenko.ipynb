```python
class BinaryTree:
    def __init__(self, *args):
        self.root = EmptyNode()

        if len(args) > 0:
            for x in args[0]:
                self.root = self.root.insert(x)
    def __repr__(self):
        return repr(self.root)
    def insert(self, value):
        self.root = self.root.insert(value)

    def __contains__(self,value):
        return value in self.root

class BinaryNode:
    def __init__(self, left, value, right):
        self.left = left
        self.value = value
        self.right = right
    def __repr__(self):
        return f'({self.left}, {self.value}, {self.right})'
    def insert(self, value):
        if '__iter__' in dir(value) and not (isinstance(value,str) and len(value)==1):
            for x in value:
                self.insert(x)
            return self
        alpha = value < self.value
        if alpha:
            self.left = self.left.insert(value)
        else:
            self.right = self.right.insert(value)
        return self

    def __contains__(self,value):
        if value == self.value:
            return True
        elif value > self.value:
            return value in self.right
        elif value < self.value:
            return value in self.left

class EmptyNode:
    def __init__(self):
        pass
    def insert(self, value):
        return BinaryNode(self, value, self)
    def __repr__(self):
        return "*"

    def __contains__(self,value):
        return False
```

## Задание 7.5. Перегрузка встроенной функции len

```python
class BinaryTree:
    def __init__(self, *args):
        self.root = EmptyNode()

        if len(args) > 0:
            for x in args[0]:
                self.root = self.root.insert(x)
    def __repr__(self):
        return repr(self.root)
    def insert(self, value):
        self.root = self.root.insert(value)
    def __contains__(self,value):
        return value in self.root

    def __len__(self):
        return len(self.root)

class BinaryNode:
    def __init__(self, left, value, right):
        self.left = left
        self.value = value
        self.right = right
    def __repr__(self):
        return f'({self.left}, {self.value}, {self.right})'
    def insert(self, value):
        if '__iter__' in dir(value) and not (isinstance(value,str) and len(value)==1):
            for x in value:
                self.insert(x)
            return self
        alpha = value < self.value
        if alpha:
            self.left = self.left.insert(value)
        else:
            self.right = self.right.insert(value)
        return self
    def __contains__(self,value):
        if value == self.value:
            return True
        elif value > self.value:
            return value in self.right
        elif value < self.value:
            return value in self.left

    def __len__(self):
        return len(self.right)+1+len(self.left)

class EmptyNode:
    def __init__(self):
        pass
    def insert(self, value):
        return BinaryNode(self, value, self)
    def __repr__(self):
        return "*"
    def __contains__(self,value):
        return False

    def __len__(self):
        return 0
```
